1.For loops - list/dictionary ke through iterate karna: 

```python
orders = [250,1200,45,3400,89,670]

#basic loop
for order in orders:
    print(f"Order value: {order}")

#enumerate() - index ke sath value bhi chahiye ho to
for idx, order in enumerate(orders):  #enumerate value ke sath sath uska index bhi print karta hai.
    print(f"Order #{idx+1}: {order}")  #kyuki python me indexing 0 se start hoti hai isliye idx ke sath +1 add kiya taki index 1 se start ho.
```

2.While loop - condition true hone tak chalta rehna:

```python
total_spend = 0 #ye ek variable banaya jis me orders ka amount judega ye badhta jaiga.
i = 0   #ye hamara counter hai jo list ke index ko track karega(hum kis order per hai)
while total_spend < 5000 and i < len(orders): #len(orders) check karta ha ki list khatam to nahi ho gyi agar sare orders kahtam ho gye to loop apne aap ruk jaiga taki error na aaye.
    total_spend += orders[i]
    i += 1
print(f"Total after {i} orders: {total_spend}")
```

3.Conditional logic loops ke andar (real business logic jaisa):

```python
for order in orders:
    if order > 1000:
        print(f"{order} -> High value order")
    elif order > 200:
        print(f"{order} -> Medium value order")
    else:
        print(f"{order} -> low value order")
```

4.Functions - reusable logic banana (yeh sabse important part hai):

```python
def classify_order(value):
    """Ek order value ko category mein classify karta hai."""
    if value > 1000:
        return "High"
    elif value > 200:
        return "Medium"
    else:
        return "Low"

# Function ko loop ke saath use karna
for order in orders:
    category = classify_order(order)
    print(f"Order {order}: {category}")
```
