1.For loops - list/dictionary ke through iterate karna: 

```python
orders = [250,1200,45,3400,89,670]

#basic loop
for order in orders:
    print(f"Order value: {order}")

#enumerate() - index ke sath value bhi chahiye ho to
for idx, order in enumerate(orders):  #enumerate value ke sath sath uska index bhi print karta hai.
    print(f"Order #{idx+1}: {order}")  #kyuki python me indexing 0 se start hoti hai isliye idx ke sath +1 add kiya taki index 1 se start ho.
```

2.While loop - condition true hone tak chalta rehna:

```python
total_spend = 0 #ye ek variable banaya jis me orders ka amount judega ye badhta jaiga.
i = 0   #ye hamara counter hai jo list ke index ko track karega(hum kis order per hai)
while total_spend < 5000 and i < len(orders): #len(orders) check karta ha ki list khatam to nahi ho gyi agar sare orders kahtam ho gye to loop apne aap ruk jaiga taki error na aaye.
    total_spend += orders[i]
    i += 1
print(f"Total after {i} orders: {total_spend}")
```

3.Conditional logic loops ke andar (real business logic jaisa):

```python
for order in orders:
    if order > 1000:
        print(f"{order} -> High value order")
    elif order > 200:
        print(f"{order} -> Medium value order")
    else:
        print(f"{order} -> low value order")
```

4.Functions - reusable logic banana (yeh sabse important part hai):

```python
def classify_order(value):
    """Ek order value ko category mein classify karta hai."""
    if value > 1000:
        return "High"
    elif value > 200:
        return "Medium"
    else:
        return "Low"

# Function ko loop ke saath use karna
for order in orders:
    category = classify_order(order)
    print(f"Order {order}: {category}")
```

5.Function jo Recency/Churn logic ka preview de (Bonus Day 35B ke liye direct foundation):

```python
def is_churned(days_since_last_order, threshold=90):
    """
    Business rule: agar customer ne threshold din se zyada 
    time se order nahi kiya, to churned maano.
    """
    if days_since_last_order > threshold:
        return 1  # churned
    else:
        return 0  # not churned

# Test karo
test_days = [30, 95, 200, 60, 15]
for days in test_days:
    status = is_churned(days)
    print(f"Days since last order: {days} -> Churned: {status}")
```

6.Function jo multiple values return kare (tuple return karke):

```python
def analyze_orders(order_list):
    """Ek list of orders leke summary stats return karta hai."""
    total = sum(order_list)
    average = total / len(order_list)
    highest = max(order_list)
    lowest = min(order_list)
    return total, average, highest, lowest

total, avg, high, low = analyze_orders(orders)
print(f"Total: {total}, Avg: {avg:.2f}, Highest: {high}, Lowest: {low}")
```

7.Nested loop — real dataset jaisa scenario (customers ke andar unke orders loop karna):

```python
customer_orders = {
    14646: [500, 1200, 300],
    13047: [90, 150],
    17850: [2000, 3400, 1800, 250]
}

for customer_id, order_list in customer_orders.items():
    total, avg, high, low = analyze_orders(order_list)
    print(f"Customer {customer_id}: Total={total}, Avg={avg:.2f}, Orders={len(order_list)}")
```
