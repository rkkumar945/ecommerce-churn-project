1.Pandas mein 2 main data structures hoti hain:

:Series — ek single column (1D), jaise ek list lekin index ke saath

:DataFrame — poori table (2D), multiple columns wala — yeh tumhara transactions data hai

1.Series banana:

```python
import pandas as pd

prices = pd.Series([6.95, 2.10, 3.75, 15.00])
print(prices)
print(type(prices))

# Custom index ke saath
prices_named = pd.Series([6.95, 2.10, 3.75], index=['Item A', 'Item B', 'Item C'])
print(prices_named)
print(prices_named['Item B'])
```

2. DataFrame manually banana (dictionary se):

```python
data = {
    'customer_id': [14646, 13047, 17850],
    'country': ['United Kingdom', 'Germany', 'France'],
    'total_spend': [5230.50, 1200.00, 800.75]
}
df = pd.DataFrame(data)
print(df)
print(type(df))
```

3. DataFrame ke basic attributes:

```python
print("Shape:", df.shape)         # (rows, columns)
print("Columns:", df.columns)
print("Index:", df.index)
print("Data types:\n", df.dtypes)
```

4. Ek column DataFrame se nikalna — yeh actually ek Series hi hoti hai:

```python
country_col = df['country']
print(country_col)
print(type(country_col))   # Series!

# Multiple columns nikalna (yeh DataFrame rehta hai)
subset = df[['customer_id', 'total_spend']]
print(type(subset))   # DataFrame
```

5. Apna already bana hua database se DataFrame load karna (jo humhare paas already hai):

```python
import sqlite3

conn = sqlite3.connect('../data/db/ecommerce.db')
df_transactions = pd.read_sql("SELECT * FROM transactions LIMIT 10", conn)
conn.close()

print(df_transactions)
print(type(df_transactions))
```

6. Series par mathematical operations (NumPy jaisa hi behavior, kyunki Pandas NumPy par built hai):

```python
spend_series = df['total_spend']
print("Sum:", spend_series.sum())
print("Mean:", spend_series.mean())
print("Max:", spend_series.max())

# Series par condition
high_spenders = spend_series > 1000
print(high_spenders)
```

7. DataFrame aur Series ka relationship samajhna — ek chhota experiment:

```python
# Har column ek Series hai
for col in df.columns:
    print(f"{col}: {type(df[col])}")

# Poora DataFrame Series ka collection hai
print(df.dtypes)
```

1.Ek Series banao 6 StockCodes ki (manually type karke), aur ek Series banao unke corresponding Prices ki. Dono ko combine karke ek naya DataFrame banao (pd.DataFrame({'stock_code': ..., 'price': ...})).

```python
import pandas as pd

# 1. 6 Stock Codes ki Series manually create karna
stock_codes = pd.Series(['RELIANCE', 'TCS', 'INFY', 'HDFCBANK', 'ICICIBANK', 'SBIN','RAJAT'])

# 2. Unke corresponding Prices ki Series manually create karna
prices = pd.Series([2950.50, 4120.00, 1650.25, 1540.00, 1120.75, 780.40,780.0])

# 3. Dono Series ko combine karke DataFrame banana
df = pd.DataFrame({
    'stock_code': stock_codes,
    'price': prices
})

print(df)
```

2.Apne data/db/ecommerce.db se customers table ko poori Pandas DataFrame mein load karo, aur uska .shape aur .dtypes print karo.

```python
import sqlite3
import pandas as pd

# 1. SQLite database se connection establish karna
conn = sqlite3.connect('../data/db/ecommerce.db')

# 2. Customers table ko pandas DataFrame mein load karna
query = "SELECT * FROM customers"
df = pd.read_sql(query, conn)

# 3. Connection close karna
conn.close()

# 4. Shape aur dtypes print karna
print("DataFrame Shape (Rows, Columns):", df.shape)
print("\nData Types of each column:\n", df.dtypes)
```
