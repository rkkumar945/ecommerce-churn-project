Note: Pandas mein 2 main tareeke hain rows/columns select karne ke:


1: .loc[] — label-based (column names, index labels use karke)


2: .iloc[] — position-based (integer position, jaise 0,1,2... — list ki tarah)

1.Data load karo

```python
import pandas as pd

df = pd.read_csv(
    '../data/raw/online_retail_II.csv',
    dtype={'Invoice': str, 'StockCode': str},
    parse_dates=['InvoiceDate']
)
```

2.Column selection — single aur multiple:

```python
# Single column (Series milegi)
prices = df['Price']
print(type(prices))

# Multiple columns (DataFrame milega)
subset = df[['Invoice', 'StockCode', 'Price']]
print(subset.head())
```

3 .iloc[] — position se rows/columns select karna:

```python
print(df.iloc[0])           # pehli row (Series)
print(df.iloc[0:5])         # pehli 5 rows (jaise list slicing)
print(df.iloc[0, 0])        # row 0, column 0 (single value)
print(df.iloc[0:5, 0:3])    # pehli 5 rows, pehle 3 columns
print(df.iloc[-1])          # aakhri row
print(df.iloc[:, 2])        # sab rows, sirf 3rd column (index 2)
```

4 .loc[] — label se rows/columns select karna:

```python
print(df.loc[0])                          # index label 0 wali row
print(df.loc[0:5])                        # NOTE: loc mein end INCLUSIVE hota hai (0 to 5 = 6 rows)
print(df.loc[0:5, ['Invoice', 'Price']])  # specific columns by name
print(df.loc[:, 'Country'])               # sab rows, sirf Country column
```

Important note: .iloc[0:5] = 5 rows (0,1,2,3,4), lekin .loc[0:5] = 6 rows (0,1,2,3,4,5) — kyunki .loc mein end label included hota hai, .iloc mein Python jaisa excluded.

5.Conditional filtering with .loc[] — bahut common pattern (boolean masking):

```python
# Sirf UK ke rows
uk_rows = df.loc[df['Country'] == 'United Kingdom']
print(uk_rows.shape)

# Multiple conditions (& = AND, | = OR) — hamesha brackets use karna
high_value = df.loc[(df['Quantity'] > 0) & (df['Price'] > 10)]
print(high_value.shape)

# Condition ke saath specific columns bhi
uk_high_value = df.loc[(df['Country'] == 'United Kingdom') & (df['Price'] > 50), ['Invoice', 'Description', 'Price']]
print(uk_high_value.head())
```

6 .loc[] se values update karna (real data cleaning mein kaam aayega):

```python
df_copy = df.copy()  # original ko change na karne ke liye copy banao
df_copy.loc[df_copy['Country'] == 'EIRE', 'Country'] = 'Ireland'  # example: naam standardize karna
print(df_copy['Country'].unique())
```

7 at[] aur iat[] — single value fast access ke liye (bahut bada data ho to):

```python
print(df.at[0, 'Price'])       # label-based single value
print(df.iat[0, 5])            # position-based single value
```
