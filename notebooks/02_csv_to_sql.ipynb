```python
import pandas as pd
import sqlite3

#Invoice or StockCode ko string rakho - warna cancellation rows ('C' prefix)
#or alphanumeric stock codes (jaise POST, DOT) corrupt ho jayenge

df = pd.read_csv(
    '../data/raw/online_retail_II.csv',
    dtype={'Invoice':str, 'StockCode': str} # ye kaam jaruri hai kyuki Invoice me 'C' prefix(cancellations) or StockCode me letters(POST,DOT,M,ADJUST) hai- agar ise int/mixed type maan liya to Day 34 wali cancellation-filtering or day32 wali cleaning steps me dikkat aaigi.    
)  #ye dtype parameter hai jo csv file ke andar batata hai ki data ko kis tarah read karna hai. is se Invoice column ko string(text) ki treat kare or StockCode column ko bhi aise hi treat kar ek kyuki is columnn me bhi POST,DOT jaise column bhi hai.


print(df.shape)
print(df.dtypes)
```

```python
#pyhton code pandas DataFrame(df) ke poore data ko SQLite database me load(save) kiya.
conn = sqlite3.connect('../data/db/ecommerce.db') #connection establish kiya.
df.to_sql('transactions', conn, if_exists='replace', index=False) #pandas ke built-in function ke through DataFrame ko direct SQL database table me convert kar deta hai. 
conn.close()
```

```python
#verify karne ke liye

conn = sqlite3.connect('../data/db/ecommerce.db')
check = pd.read_sql("SELECT COUNT(*)  as total_rows FROM transactions", conn) #ye pd.read ek pandas function hai jo SQL query chalakar result ko seedha ek pandas DataFrame me change kar deta hai
print(check)# total rows ko count karke bata dega.

sample = pd.read_sql("SELECT * FROM transactions LIMIT 5",conn) # sabhi columns select karke pehli 5 rows print ki.
print(sample)
conn.close()
```
