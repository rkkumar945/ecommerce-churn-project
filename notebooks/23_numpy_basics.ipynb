2.Array vs List - speed aur behaviour ka fark:

```python
# List mein har element pe operation karne ke liye loop chahiye
prices_list = [10, 20, 30]
discounted_list = [p * 0.9 for p in prices_list]   # list comprehension
print(discounted_list)

# NumPy array mein directly operation ho jaata hai (vectorized)
prices_array = np.array([10, 20, 30])
discounted_array = prices_array * 0.9
print(discounted_array)
```

Note - NumPy arrays par mathematical operations element-wise automatically apply hote hain, aur yeh Python lists se bahut zyada fast hota hai bade datasets par (kyunki NumPy internally C mein likha hai).

3.2D Arrays (matrix jaisa — jaise ek chhota transactions table):

```python
# Rows = orders, Columns = [Quantity, Price]
transactions_sample = np.array([
    [12, 6.95],
    [48, 2.10],
    [24, 3.75],
    [6, 15.00]
])
print(transactions_sample.shape)   # (4, 2)
print(transactions_sample[0])      # pehli row
print(transactions_sample[:, 0])   # sirf pehla column (Quantity)
print(transactions_sample[:, 1])   # sirf dusra column (Price)
```

4.Basic operations — sum, mean, max, min:

```python
quantities = transactions_sample[:, 0]
prices = transactions_sample[:, 1]

print("Total quantity:", np.sum(quantities))
print("Average price:", np.mean(prices))
print("Max price:", np.max(prices))
print("Min quantity:", np.min(quantities))
print("Std deviation of prices:", np.std(prices))
```

5.Revenue calculate karna — element-wise multiplication (Quantity × Price, jaisa hum SQL mein karte the):

```python
revenue_per_order = quantities * prices
print(revenue_per_order)
print("Total revenue:", np.sum(revenue_per_order))
```

6.. Boolean indexing/filtering (bahut important pattern — Pandas mein bhi yehi concept use hoga):

```python
# Sirf woh orders dikhao jinka revenue > 50
high_value_mask = revenue_per_order > 50
print(high_value_mask)                      # True/False array
print(revenue_per_order[high_value_mask])   # sirf True wale values
```

Practice questions.

1.Ek NumPy array banao 20 random order values ka (np.random.randint(10, 5000, size=20)), phir uska mean, median (np.median()), aur standard deviation nikalo.

```python
import numpy as np

# 10 se 5000 ke beech 20 random orders ka NumPy array banana
orders_array = np.random.randint(10, 5000, size=20)
print(f"Orders Array:\n{orders_array}\n")

# 2. Mean (Average) nikalna
mean_value = np.mean(orders_array)

# 3. Median (Beech ki value) nikalna
median_value = np.median(orders_array)

# 4. Standard Deviation (Variation / Failaav) nikalna
std_value = np.std(orders_array)

# Results print karna
print(f"Mean (Average): {mean_value:.2f}")
print(f"Median: {median_value}")
print(f"Standard Deviation: {std_value:.2f}")
```

2.Boolean indexing use karke un values ko nikalo jo array ke mean se zyada hain.

```python
# Mean se zyada wali values nikalna (Boolean Indexing)
high_orders = orders_array[orders_array > mean_value]

print(f"Mean (Average) tha: {mean_value:.2f}")
print(f"Mean se zyada values: {high_orders}")
```

3.Do arrays banao — quantity aur price (5-5 elements ke) — aur unka element-wise product (revenue) nikalo, phir total revenue print karo.

```python
import numpy as np

# 1. Quantity aur Price ke 5-5 elements wale do arrays banana
quantity = np.array([2, 5, 1, 4, 3])
price = np.array([100, 250, 1200, 50, 500])

# 2. Element-wise product (Revenue) nikalna
# NumPy seedhe corresponding elements ko multiply kar deta hai (jaise quantity[0] * price[0])
revenue = quantity * price

# 3. Total revenue nikalna (saare revenues ka sum)
total_revenue = np.sum(revenue)

# Results print karna
print(f"Quantity: {quantity}")
print(f"Price: {price}")
print(f"Revenue (Item-wise): {revenue}")
print(f"Total Revenue: {total_revenue}")
```
