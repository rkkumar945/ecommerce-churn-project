1.Data load karo.

```python
import pandas as pd

df = pd.read_csv(
    '../data/raw/online_retail_II.csv',
    dtype={'Invoice': str, 'StockCode': str},
    parse_dates=['InvoiceDate']
)
```

1.Basic single-condition filtering (jo Day 29 mein preview dekha, ab detail mein):

```python
# Sirf woh rows jinka bill (Quantity * Price) 500 se zyada hai
df['LineTotal'] = df['Quantity'] * df['Price']
high_bills = df[df['LineTotal'] > 500]
print(high_bills.shape)
print(high_bills[['Invoice', 'Description', 'LineTotal']].head())
```

2.Multiple conditions — AND (&), OR (|), NOT (~):

```python
# AND — dono conditions true honi chahiye
uk_high_value = df[(df['Country'] == 'United Kingdom') & (df['LineTotal'] > 500)]
print(uk_high_value.shape)

# OR — koi bhi ek condition true ho
germany_or_france = df[(df['Country'] == 'Germany') | (df['Country'] == 'France')]
print(germany_or_france.shape)

# NOT — condition ka ulta
not_uk = df[~(df['Country'] == 'United Kingdom')]
print(not_uk.shape)
```

Note: Python ke and/or yahan kaam nahi karte — Pandas mein &, |, ~ use karna zaroori hai, aur har condition ko bracket mein rakhna zaroori hai warna error aayega.

3.isin() — multiple specific values check karna ek saath (Day 29 ka preview, ab detail):

```python
selected_countries = df[df['Country'].isin(['United Kingdom', 'Germany', 'France'])]
print(selected_countries['Country'].unique())

# Opposite — inn countries ko chhod kar
excluded = df[~df['Country'].isin(['United Kingdom', 'Germany', 'France'])]
print(excluded['Country'].unique())
```

4. String-based filtering — .str accessor (text columns par condition):

```python
# Description mein "CHRISTMAS" word wale products
christmas_items = df[df['Description'].str.contains('CHRISTMAS', na=False)]
print(christmas_items.shape)
print(christmas_items['Description'].unique()[:10])

# Invoice 'C' se start hone wale (cancellations — Day 34 mein detail se cleaning karenge)
cancellations = df[df['Invoice'].str.startswith('C', na=False)]
print(cancellations.shape)
```

Note: na=False zaroori hai — agar value NaN hai to .str.contains() error dega, na=False use karne se NaN wali rows automatically False (exclude) ho jaati hain.

5.Date-based filtering (Datetime column par condition):

```python
# Sirf December 2011 ke orders
dec_2011 = df[(df['InvoiceDate'] >= '2011-12-01') & (df['InvoiceDate'] < '2012-01-01')]
print(dec_2011.shape)

# `.dt` accessor se specific parts nikalna
df['Year'] = df['InvoiceDate'].dt.year
df['Month'] = df['InvoiceDate'].dt.month
recent_data = df[df['Year'] == 2011]
print(recent_data.shape)
```

6.query() method — SQL jaisa readable syntax (alternative tareeka):

```python
high_value_query = df.query('LineTotal > 500 and Country == "United Kingdom"')
print(high_value_query.shape)
```

note: Yeh exact same result dega jaisa & wala version — bas syntax zyada SQL jaisa/readable lagta hai.

7.Filtering + aggregation combine karna (chained operations):

```python
# High-value UK orders ka total revenue
total_uk_high_value = df[(df['Country'] == 'United Kingdom') & (df['LineTotal'] > 500)]['LineTotal'].sum()
print(f"Total high-value UK revenue: {total_uk_high_value:.2f}")
```
