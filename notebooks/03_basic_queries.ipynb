```python
import pandas as pd
import sqlite3

conn = sqlite3.connect('../data/db/ecommerce.db')
```

```python
#SELECT - specific columns dekhna
q1 = pd.read_sql("SELECT Invoice, StockCode, Description, Quantity, Price FROM transactions LIMIT 10", conn) #ye last me conn likhna jaruri hai kyuki pandas(python) ko ye batana jaruri hai ki hamara database(ecommerce.db) computer me kha save hai. 
print(q1)
```

```python
#1.WHERE - filtering practice
#sirf UK ke customers
q2 = pd.read_sql("SELECT * FROM transactions WHERE Country = 'United Kingdom' LIMIT 10", conn)


#Quantity 0 se zyada or Price 5 se zyada
q3 = pd.read_sql("SELECT * FROM transactions WHERE Quantity > 0 AND Price > 5 LIMIT 10", conn)


#Cnacellations dhoondo (Invoice 'C' se start)
q4 = pd.read_sql("SELECT * FROM transactions WHERE Invoice LIKE 'C%' LIMIT 10", conn) # LIKE 'C%' ka use isliye hota hai taki aise Invoices dhundo jo C se start hote hai or % ek wildcard hai jo batata hai ki 'C' ke baad chahe kuch bhi likha ho chalega bas column C se start hona chaiye
print(q4.shape)


#Missing Customer ID wale rows
q5 = pd.read_sql("SELECT * FROM transactions WHERE \"Customer ID\" IS NULL LIMIT 10", conn)#kyuki Customer ID column name ke buch me space hai to isliye " " ye lagaya gya hai. 
```

```python
#ORDER BY - sorting practice

#sabse mehenge items
q6 = pd.read_sql("SELECT Description, Price FROM transactions ORDER BY Price DESC LIMIT 10", conn)#DESC = bade se chota


#sabse purani transactions
q7 = pd.read_sql("SELECT Invoice, InvoiceDate FROM transactions ORDER BY InvoiceDate ASC LIMIT 10", conn)#agar hum ASC bhi na lagaye to bhi SQL default roop me ascending order hi manta hai.


#Country ke hisaab se alphabetically
q8 = pd.read_sql("SELECT DISTINCT Country FROM transactions ORDER BY Country ASC", conn) #DISTINICT keyword ka use duplicate(repeated) values ko hatane ke liye hota hai.
print(q8)
```

```python
#WHERE + ORDER BY ek sath practice

#UK ke bahar ke customers, sabse zyada Quantity wale top 10
q9 = pd.read_sql("""
    SELECT Country, Description, Quantity, Price
    FROM transactions
    WHERE Country != 'United Kingdom' AND Quantity > 0 
    ORDER BY Quantity DESC
    LIMIT 10
""", conn)

print(q9)
```
