1.Mathematical functions — NumPy ke built-in math operations:

```python
import numpy as np

order_values = np.array([250, 1200, 45, 3400, 89, 670])

print("Square root:", np.sqrt(order_values))
print("Log (natural):", np.log(order_values))
print("Exponential:", np.exp(np.array([1, 2, 3])))  # chhota example, bade numbers overflow karte hain
print("Power (squared):", np.power(order_values, 2))
print("Absolute value:", np.abs(np.array([-50, 30, -20, 10])))
```

2.Rounding functions — real data cleaning mein kaam aayenge:

```python
prices = np.array([6.9499, 2.1032, 3.7567, 15.0011])

print("Round to 2 decimals:", np.round(prices, 2))
print("Floor (round down):", np.floor(prices))
print("Ceil (round up):", np.ceil(prices))
```

3.Cumulative functions — running totals (Day 14 wala LAG/running-total concept, ab NumPy mein):

```python
daily_revenue = np.array([500, 300, 700, 200, 900])

print("Cumulative sum:", np.cumsum(daily_revenue))
print("Cumulative product:", np.cumprod(np.array([1.1, 1.05, 0.95, 1.2])))  # jaise growth rate compound karna
```

4.Percentile aur Quantile — Day 13 wala NTILE concept, ab NumPy mein:

```python
customer_spends = np.array([100, 250, 400, 800, 1200, 1500, 2000, 5000, 8000, 12000])

print("25th percentile:", np.percentile(customer_spends, 25))
print("50th percentile (median):", np.percentile(customer_spends, 50))
print("75th percentile:", np.percentile(customer_spends, 75))
print("90th percentile:", np.percentile(customer_spends, 90))
```

```python
#Yeh exactly wahi concept hai jo Month 3 mein IQR (Interquartile Range) se outliers dhoondne ke liye use hoga — abhi bas function samajh lo.
```

5.Correlation — do variables ka relationship (Month 3 stats ka preview):

```python
quantity = np.array([12, 48, 24, 6, 100, 30])
price = np.array([6.95, 2.10, 3.75, 15.00, 1.20, 5.50])

correlation_matrix = np.corrcoef(quantity, price)
print(correlation_matrix)
print("Correlation coefficient:", correlation_matrix[0, 1])
```

```python
#Value -1 se +1 ke beech hoti hai — negative matlab ek badhne se dusra ghatta hai (jaise yahan zyada quantity wale orders mein per-unit price kam hoti hai — bulk discount jaisa pattern).
```

6.np.where() — conditional logic (Pandas mein bhi bahut use hoga):

```python
revenue = quantity * price
category = np.where(revenue > 50, 'High', 'Low')
print(revenue)
print(category)
```

7.NaN handling (missing values ka preview — Month 2 Week 7 mein detail se aayega):

```python
data_with_missing = np.array([100, np.nan, 250, np.nan, 400])

print("Is NaN:", np.isnan(data_with_missing))
print("Mean ignoring NaN:", np.nanmean(data_with_missing))
print("Sum ignoring NaN:", np.nansum(data_with_missing))
```

1.customer_spends array ka IQR nikaalo (Q3 - Q1, yani 75th percentile minus 25th percentile) — yeh outlier detection ka core formula hai.

```python
customer_spends = np.array([100,250,400,800,1200,1500,2000,5000,8000,12000])

q1 = np.percentile(customer_spends,25)
q2 = np.percentile(customer_spends,75)


iqr = q2 - q1

print("25th percentile (q1):",q1)
print("75th percentile (q2):",q2)
print("IQR",iqr)
```

2.Ek array banao 15 random values ka jisme kuch np.nan bhi ho, phir nanmean() aur normal mean() ka result compare karo — dekho normal mean kya deta hai jab NaN ho (hint: NaN aata hai).

```python
# 1. 15 random values ki array banate hain (0 se 100 ke beech)
np.random.seed(42) # Taaki har baar same random numbers aayein
data = np.random.randint(10, 100, size=15).astype(float)

# 2. Kuch positions par purposely np.nan (missing value) dal dete hain
data[3] = np.nan
data[7] = np.nan
data[12] = np.nan

print("Array with NaNs:\n", data)

# 3. Normal mean aur nanmean calculate karna
normal_mean = np.mean(data)
clean_mean = np.nanmean(data)

print("\n Comparison-- ")
print("Normal Mean (np.mean):", normal_mean)
print("Mean ignoring NaN (np.nanmean):", clean_mean)
```
