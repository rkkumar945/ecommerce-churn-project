```python
import pandas as pd

df = pd.read_csv(
    '../data/raw/online_retail_II.csv',
    dtype={'Invoice': str, 'StockCode': str},
    parse_dates=['InvoiceDate']
)
```

Section A: Loading & Overview (Day 26-28)

1.Dataset ko sahi dtype aur parse_dates ke saath load karo, phir .info() aur .describe() dono ka output dikhao — in dono outputs se 3 observations apne shabdon mein likho.

```python
df = pd.read_csv('../data/raw/online_retail_II.csv', parse_dates=['InvoiceDate'])

print("Information--")
print(df.info())

print("\n Describe--")
print(df.describe())
```

2.Country column par value_counts() lagao aur top 5 countries dikhao, saath mein unka percentage share bhi (normalize=True use karke).

```python
top_countries = df['Country'].value_counts(normalize=True).head(5) * 100
print("Top 5 countries with Percentage share:")
print(top_countries)
```

Section B: Indexing & Slicing (Day 29)

3.loc[] use karke sirf Spain aur Portugal ke rows nikaalo (hint: .isin()), aur sirf Invoice, Description, Price columns dikhao.

```python
result_q3 = df.loc[
    df['Country'].isin(['Spain', 'Portugal']),
    ['Invoice', 'Description', 'Price']
]

print("shape:",result_q3.shape)
print(result_q3.head())
```

4.iloc[] use karke DataFrame ki last 15 rows aur pehle 5 columns nikaalo.

```python
result_q4 = df.iloc[-15:, :5]
print(result_q4)
```

Section C: Conditional Filtering (Day 30)

5.Ek naya column LineTotal banao (Quantity * Price), phir un rows ko filter karo jinka LineTotal > 1000 aur Country == 'Netherlands'.

```python
df['LineTotal'] = df['Quantity'] * df['Price']

result_q5 = df[(df['LineTotal'] > 1000) & (df['Country'] == 'Netherlands')]
print("shape:", result_q5.shape)
print(result_q5.head())
```

6.str.contains() use karke Description mein "BAG" word wale products dhoondo, aur unme se top 5 sabse zyada bikne wale (Quantity sum ke hisaab se) products nikaalo.

```python
bag_df = df[df['Description'].str.contains('BAG', case=False, na=False)]
top_bags = bag_df.sort_values(by='Quantity', ascending=False).head(5)

print(top_bags[['Description', 'Quantity', 'Price', 'Country']])
print("shape:", top_bags.shape)
```

7.Invoice 'C' se start hone wali (cancellations) rows ko filter karo, aur query() method se wahi result dobara nikaalo — confirm karo dono same shape de rahe hain.

```python
# Method 1: Boolean Indexing
c_filter1 = df[df['Invoice'].astype(str).str.startswith('C')]

# Method 2: query() method
c_filter2 = df.query("Invoice.str.startswith('C', na=False)")

print("Boolean Indexing Shape:", c_filter1.shape)
print("Query Method Shape:", c_filter2.shape)
print("Dono same hain kya?", c_filter1.shape == c_filter2.shape)
```

Section D: Combine Everything

8.Ek query banao jo: 2011 ke saal ke, United Kingdom ke, aur LineTotal > 500 wale orders nikaale — phir unka total revenue aur average revenue print karo.

```python
mask = (df['InvoiceDate'].dt.year == 2011) & (df['Country'] == 'United Kingdom') & (df['LineTotal'] > 500)
filtered_uk = df[mask]

total_revenue = filtered_uk['LineTotal'].sum()
avg_revenue = filtered_uk['LineTotal'].mean()

print(f"Shape: {filtered_uk.shape}")
print(f"Total Revenue: {total_revenue:,.2f}")
print(f"Average Revenue: {avg_revenue:,.2f}")
```

9.Description column mein missing values wale rows nikaalo, aur check karo — kya inka StockCode kisi specific pattern (jaise POST, DOT, M) follow karta hai ya random hai?

```python
missing_desc = df[df['Description'].isnull()]

print("Missing Description Rows Shape:", missing_desc.shape)
print("\nTop StockCodes for missing descriptions:")
print(missing_desc['StockCode'].value_counts().head(10))
```

Section E: Conceptual

10.Apne shabdon mein (2-3 lines) explain karo: .loc[] aur .iloc[] mein kya fark hai, aur ek real example do jaha tum .loc[] use karoge aur ek jaha .iloc[] use karoge.

Ans: 1..loc[] (Label-based indexing): Isme hum DataFrame ke row aur column ke naam/labels (jaise index labels ya column headers) pass karte hain. Yeh inclusive of end boundary hota hai.


Real-life example: 
                   Jab humein specific rows chahiye jinka index label koi date ho (df.loc['2011-12-01']) ya specific columns select karne hon (df.loc[:, ['Invoice', 'Price']]).


2.iloc[] (Integer position-based indexing): Isme hum Python ke standard zero-based integer positions (numbers) use karte hain. Yeh exclusive of end boundary hota hai (slicing ke waqt).


Real-life example:
                   Jab humein DataFrame ki shuruat ki rows aur columns nikaalne hon, jaise pehli 10 rows aur pehle 3 columns (df.iloc[:10, :3]).
