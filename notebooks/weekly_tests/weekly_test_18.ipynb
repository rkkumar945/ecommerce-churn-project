**Section F: Debugging challenge (45 minute)**

Neeche ki file `scratch/buggy_app.py` mein save karo (`scratch/` git-ignored hai). Ye jaanbujhkar buggy hai, aur apne aap mein poori hai (tumhare data par depend nahi karti).
```python
import time

import numpy as np
import pandas as pd
import streamlit as st
from sklearn.linear_model import LogisticRegression

FEATURES = ['recency', 'frequency', 'spend']


def load_model():
    time.sleep(2)                                   # bhaari model load ka simulation
    rng = np.random.default_rng(0)
    X = pd.DataFrame(rng.normal(size=(300, 3)), columns=FEATURES)
    y = (X['recency'] - X['frequency'] + rng.normal(size=300) > 0).astype(int)
    return LogisticRegression().fit(X, y)


model = load_model()

st.title('Buggy churn app')

customer_id = st.text_input('Customer ID')
recency = st.slider('Recency', -3.0, 3.0, 0.0)
frequency = st.slider('Frequency', -3.0, 3.0, 0.0)
spend = st.slider('Spend', -3.0, 3.0, 0.0)

history = []

if st.button('Score'):
    cid = int(customer_id)
    row = pd.DataFrame([{'spend': spend, 'recency': recency, 'frequency': frequency}])
    p = model.predict_proba(row)[0, 1]
    history.append((cid, round(float(p), 2)))
    st.metric('Score', f'{p:.2f}')

st.write('History:', history)

if st.checkbox('Show details'):
    st.slider('Recency', -3.0, 3.0, 0.0)
```
