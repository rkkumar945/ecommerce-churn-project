1.Data load karo (poora dataset, jaisa Day 27 mein seekha):

```python
import pandas as pd

df = pd.read_csv(
    '../data/raw/online_retail_II.csv',
    dtype={'Invoice': str, 'StockCode': str},
    parse_dates=['InvoiceDate']
)
print(df.shape)
```

2.head() aur tail() — pehli/aakhri rows dekhna:

```python
print(df.head())        # default 5 rows
print(df.head(10))      # custom count
print(df.tail())        # aakhri 5 rows
print(df.sample(5))     # random 5 rows — zyada useful hai kabhi kabhi (head sirf shuru dikhata hai, bias ho sakta hai)
```

3.info() — poori DataFrame ka summary (data types, non-null counts, memory):

```python
df.info()
```

4.describe() — numerical columns ka statistical summary:

```python
print(df.describe())
```

Note = count, mean, std, min, 25%, 50% (median), 75%, max — yeh sab Day 25 mein NumPy se manually nikale the, describe() sab ek saath de deta hai.

5.describe() categorical (text) columns ke liye bhi:

```python
print(df.describe(include='object'))
```

Note = Yeh count, unique, top (sabse common value), freq (uski frequency) dikhayega — jaise Country column mein kaunsa country sabse zyada hai.

6.describe() poore dataset (numeric + categorical dono) ke liye:

```python
print(df.describe(include='all'))
```

7.Individual column ki jhalak — unique(), nunique(), value_counts():

```python
print("Unique countries:", df['Country'].nunique())
print(df['Country'].unique())          # saari unique values ki list
print(df['Country'].value_counts())    # har value ka count, sorted
print(df['Country'].value_counts(normalize=True) * 100)  # percentage mein
```

9.df.isnull().sum() — quick missing value check (Day 32 mein detail se karenge, aaj sirf overview):

```python
print(df.isnull().sum())
```
