API = API (Application Programming Interface) ek tarika hai jisse do alag software aapas mein data      exchange karte hain — jaise weather app humhare phone par internet se live weather data laata hai, wahan background mein ek API call ho rahi hoti hai.

1.Ek simple free public API se data fetch karna — Joke API:

```python
import requests
import pandas as pd

response = requests.get('https://official-joke-api.appspot.com/random_joke')
print(response.status_code)   # 200 matlab success
print(response.json())        # response ko JSON format mein dekho



#dataframe me dekhna
df_live = pd.json_normalize([response.json()])
df_live.head()
```

2.Weather API try karo (free, no API key chahiye — Open-Meteo):

```python
# Delhi ke coordinates use karke
response = requests.get(
    'https://api.open-meteo.com/v1/forecast',
    params={
        'latitude': 28.6139,
        'longitude': 77.2090,
        'current_weather': True
    }
)
data = response.json()
print(data)
print("Current temperature:", data['current_weather']['temperature'])
```

params dictionary se hum API ko batate hain ki kaunsa specific data chahiye — yeh API request ka hi hissa hai (query parameters).

3.Response ke important parts samajhna:

```python
response = requests.get('https://official-joke-api.appspot.com/random_joke')

print("Status Code:", response.status_code)   # 200=OK, 404=Not Found, 500=Server Error
print("Headers:", response.headers['Content-Type'])
print("Raw text:", response.text[:100])       # raw string form
print("Parsed JSON:", response.json())         # Python dict form
```

4.. Error handling — agar API fail ho jaye to crash na ho:

```python
try:
    response = requests.get('https://official-joke-api.appspot.com/random_joke', timeout=5)
    response.raise_for_status()  # agar error status ho to exception raise karega
    data = response.json()
    print(f"Joke: {data['setup']} -- {data['punchline']}")
except requests.exceptions.RequestException as e:
    print(f"API call failed: {e}")
```

2.Open-Meteo API mein apne shehar (Meerut) ke coordinates daal kar current temperature nikaalo.

```python
# Meerut ke coordinates
latitude = 28.9845
longitude = 77.7064

# Open-Meteo API URL (current temperature ke liye)
url = f"https://api.open-meteo.com/v1/forecast?latitude={latitude}&longitude={longitude}&current=temperature_2m"

# API request bhejte hain
response = requests.get(url)

# Check karte hain ki request successful hai ya nahi
if response.status_code == 200:
    data = response.json()
    
    # Current temperature nikalna
    current_temp = data["current"]["temperature_2m"]
    temp_unit = data["current_units"]["temperature_2m"]
    
    print("Status Code:", response.status_code)
    print(f"Meerut ka Current Temperature: {current_temp} {temp_unit}")
else:
    print("API call fail ho gayi, Status Code:", response.status_code)
```
