JSON format = JSON (JavaScript Object Notation) data likhne ka ek universal format hai — keys aur values ke jodon mein, jaise Python dictionary. Zyadatar APIs isi format mein data bhejti/leti hain, isliye JSON samajhna zaroori hai.

1.Python ka json module — dictionary ko JSON string mein convert karna:

```python
import json

customer = {
    "customer_id": 14646,
    "country": "United Kingdom",
    "total_spend": 5230.50,
    "is_active": True,
    "past_orders": [250, 1200, 800]
}

# Dictionary -> JSON string (serialization)
json_string = json.dumps(customer)
print(json_string)
print(type(json_string))   # str

# Sundar formatting ke saath (indent se readable banta hai)
json_pretty = json.dumps(customer, indent=4)
print(json_pretty)
```

2.JSON string -> Python dictionary wapas convert karna (deserialization):

```python
raw_json = '{"customer_id": 14646, "country": "United Kingdom", "total_spend": 5230.50}'
parsed_data = json.loads(raw_json)
print(parsed_data)
print(type(parsed_data))   # dict
print(parsed_data['country'])
```

3.JSON file save/read karna (disk par):

```python
# Save karna
with open('../data/sample_customer.json', 'w') as f:
    json.dump(customer, f, indent=4)

# Read karna
with open('../data/sample_customer.json', 'r') as f:
    loaded_customer = json.load(f)
print(loaded_customer)
```

4.. Nested JSON — real APIs mein aisa hi complex data milta hai (jaise Day 25B ka weather response):

```python
nested_data = {
    "customer": {
        "id": 14646,
        "country": "United Kingdom"
    },
    "orders": [
        {"invoice": "489434", "value": 250.50},
        {"invoice": "489521", "value": 1200.00}
    ],
    "churn_prediction": {
        "probability": 0.23,
        "risk_level": "Low"
    }
}

# Nested values access karna
print(nested_data['customer']['country'])
print(nested_data['orders'][0]['invoice'])
print(nested_data['churn_prediction']['probability'])

# Loop se saare orders print karna
for order in nested_data['orders']:
    print(f"Invoice {order['invoice']}: ${order['value']}")
```

5.API response ko JSON se directly kaam mein laana (Day 25B/25C se connect karke):

```python
import requests

response = requests.get('https://jsonplaceholder.typicode.com/users/1')
user_data = response.json()   # yeh automatically JSON ko dict mein convert kar deta hai

print("Name:", user_data['name'])
print("Email:", user_data['email'])
print("City:", user_data['address']['city'])   # nested field
```

6.JSON aur Pandas — DataFrame ko JSON mein convert karna (Month 2 Week 6 mein kaam aayega):

```python
import pandas as pd

df = pd.DataFrame({
    'customer_id': [1, 2, 3],
    'total_spend': [1200, 3400, 800]
})
json_output = df.to_json(orient='records', indent=2)
print(json_output)
```

1.Apna khud ka nested dictionary banao (jaise ek "churn_prediction_request" structure — customer info + recent orders list), json.dumps() se print karo indent ke saath.

```python
import json

churn_prediction_request = {
    "request_id": "REQ-98765",
    "customer_info": {
        "customer_id": 14646,
        "country": "United Kingdom",
        "account_age_months": 24
    },
    "recent_orders": [
        {"order_id": "ORD-101", "amount": 250.50},
        {"order_id": "ORD-102", "amount": 1200.00},
        {"order_id": "ORD-103", "amount": 800.00}
    ],
    "model_version": "v2.1"
}

# JSON string mein convert karke sundar formatting ke sath print karna
json_output = json.dumps(churn_prediction_request, indent=4)
print(json_output)
```

2.jsonplaceholder.typicode.com/users (poori list) se GET request karo, aur har user ka name aur company.name (nested field) print karo loop use karke.

```python
# Poori list fetch kar rahe hain
response = requests.get('https://jsonplaceholder.typicode.com/users')
users = response.json()   # Yeh ek list of dictionaries return karega

# Loop chala kar har user ka name aur nested company name print karna
for user in users:
    name = user['name']
    company_name = user['company']['name']
    print(f"Name: {name} | Company: {company_name}")
```
