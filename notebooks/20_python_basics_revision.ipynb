2.Dictionaries - key-value pairs(jaise ek customer record):

```python
#dictionary banai
customer = {
    'customer_id': 14646,
    'country':'United Kingdom',
    'total_spend':5230.50
}
print(customer['country'])
customer['is_active'] = True    #new key value add karna.
print(customer.keys())
print(customer.values())
print(customer.items())

#loop karke print karna
for key,  value in  customer.items(): #dictionary ke sare data ko key or value ke pairs me bahar lata hai. key,value loop se key me pehli chij cahli jati hai or value me dusri chij chali jati hai.
    print(f"{key}: {value}")  #f-string ek method hai jis se hum  variables ko seedhe text ke andar likh sakte hai. yeh key or value ki jagah unka real data print kar deta hai.
```

3.Tuples - ordered,immutable(change nahi ho sakti):

```python
coordinates = (28.6139,77.2090) #jaise (latitude, longitude)
#coordinates[0] = 30 #yeh ERROR dega, tuple immutable hai
lat,lon = coordinates #unpacking , tuple me hum ek hi line me unko alag-alag variables (lat or lon) me daal sakte hai.
print(f"Lat: {lat}, Lon: {lon}")
```

4.Sets - unique values, unordered(duplicates automatically hat jaate hai):

```python
all_countries = ['UK', 'Germany', 'France', 'UK', 'Germany', 'Spain']
unique_countries = set(all_countries)
print(unique_countries)
print(len(unique_countries))

# Set operations
set_a = {'UK', 'Germany', 'France'}
set_b = {'Germany', 'France', 'Spain'}
print(set_a & set_b)   # intersection (common)
print(set_a | set_b)   # union (sab milakar)
print(set_a - set_b)   # difference (A mein hai, B mein nahi)
print(set_b - set_a)   # difference (B mein hai, A mein nahi)
```

```python
#Real use-case: Tumhare dataset mein set(df['Country']) se turant pata chal jaata hai kitne unique countries hain — humne yeh SQL mein DISTINCT se kiya tha, Python mein set() se hota hai.
```

5.Practice challenge — inn concepts ko combine karke ek chhota churn-relevant example:

```python
# Ek dictionary of lists banao — har country ke top 3 customers ki list
country_top_customers = {
    'United Kingdom': [17850, 13047, 12583],
    'Germany': [12662, 12471, 12681],
    'France': [12583, 12662, 14688]
}

for country, customers in country_top_customers.items():
    print(f"{country}: {len(customers)} top customers -> {customers}")

# Kaunse customer IDs multiple countries mein common hain (set intersection use karke)
uk_set = set(country_top_customers['United Kingdom'])
germany_set = set(country_top_customers['Germany'])
common = uk_set & germany_set
print(f"Common customers: {common}")
```

2.Ek dictionary banao jisme key = Country, value = us country ka average order value (manually 5 countries ke liye dummy data se) — phir sabse zyada avg wale country ka naam print karo.

```python
data = {
    'UK': 146.46,
    'Germany': 520.10,
    'France': 5230.50,
    'India': 714.8,
    'USA': 525.90
}
highest_country = max(data, key=data.get) #ye function batata hai ki comparison key ke base per nahi value ke base per karo.

print(f"highest average ordervalue: {highest_country}")
print(f"average value: {data[highest_country]}")
```
