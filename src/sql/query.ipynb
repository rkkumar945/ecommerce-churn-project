```python
#FULL OUTER JOIN agar SQLite3.39+ hai)
import pandas as pd
import sqlite3

conn = sqlite3.connect('../../data/db/ecommerce.db')

q1 = pd.read_sql("""
    SELECT r.country, c."Customer ID"
    FROM countries_ref r
    FULL OUTER JOIN customers c ON r.Country = c.Country
""",conn) #

print(q1)
```

```python
#03:CROSS JOIN - har row ko har row se combine karna(cartesion product).
#chota sa example: 3 countries * sabhi unique StockCodes ka combination (sirf demo ke liye, choti sample lo warna rows explode ho jaigi)
import pandas as pd
import sqlite3
sample_products = pd.read_sql("SELECT DISTINCT StockCode FROM transactions LIMIT 5", conn)
sample_products.to_sql('sample_products', conn, if_exists='replace', index=False)


q2 = pd.read_sql("""
    SELECT r.Country, p.StockCode
    FROM countries_ref r
    CROSS JOIN sample_products p
""",conn)

print(q2)
print(q2.shape)
```
